## The "Seven Step Process" for using `solve_ivp`

1. Make sure you have a clear set of ODEs where each variable has it's own derivative equation
2. Define a function that returns the value of each derivative using equations and the appropriate variables and model parameters.
3. Create a variable that defines/stores your initial conditions
4. Create a variable that contains all of the time values you want to use for evaluating your model
5. Call `solve_ivp` to get the solution, making sure to pass in model parameters as necessary
6. Unpack the solution into separate variables
7. Visualize and interpret the results

```python
def SIZR(t, y, b, d, a, g, k, r):
        Si = y[0]
        Ii = y[1]
        Zi = y[2]
        Ri = y[3]
        # the model equations (see Munz et al. 2009)
        f0 = b - a*Si*Zi - d*Si
        f1 = a*Si*Zi - r*Ii - d*Ii
        f2 = r*Ii + g*Ri - k*Si*Zi
        f3 = d*Si + d*Ii + k*Si*Zi - g*Ri
        return [f0, f1, f2, f3]
```
