## Let's generate our ballots!

```python
# Generating the data table
import numpy as np
import random
ballot_list=[]
for i in range(29):
    ballot_list.append(['a','b','c'])
for i in range(1):
    ballot_list.append(['a','c','b'])
for i in range(23):
    ballot_list.append(['b','a','c'])
for i in range(3):
    ballot_list.append(['b','c','a'])
for i in range(26):
    ballot_list.append(['c','b','a'])
for i in range(18):
    ballot_list.append(['c','a','b'])
# make the data into a 2D numpy array    
ballot_list=np.array(ballot_list)
# use np.random.shuffle to mix up the order just for fun and authenticity
np.random.shuffle(ballot_list)
print(ballot_list[:5])
```

# Let's take a look at couple possible solutions for the plurality voting

```python
## Option 1, go through every ballot with a loop and see who was selected
def plurality_1(bl):
    candidate_count= np.zeros(3)
    for i in range(len(bl)):
        if bl[i,0] == 'a':
            candidate_count[0]+=1
        elif bl[i,0]=='b':
            candidate_count[1]+=1
        else:
            candidate_count[2]+=1
    print("First choice votes for each candidate:\n",candidate_count)
    return np.argmax(candidate_count)
```

```python
print("Which candidate won?",plurality_1(ballot_list))
```

```python
## Option 2: create an array of all the first choice votes, then make a mask
def plurality_2(bl):
    first_choice_votes=np.array(bl[:,0])
    candidate_count= np.zeros(3)
    candidate_count[0]=len(first_choice_votes[first_choice_votes == 'a'])
    candidate_count[1]=len(first_choice_votes[first_choice_votes == 'b'])
    candidate_count[2]=len(first_choice_votes[first_choice_votes == 'c'])
    print("First choice votes for each candidate:\n",candidate_count)
    return np.argmax(candidate_count)
```

```python
print("Which candidate won?",plurality_2(ballot_list))
```

## What if we want to know the fraction of people that chose a candidate _last_?
    

```python
def lastplace(bl, c):
    # We use array indexing to get the last place votes
    # Then we create mask for those values that match the candidate
    # We sum that mask because "True" is equal to "1"
    # and then divide the result by the total number of ballots
    frac_last = (bl[:,-1] == c).sum() / len(ballot_list)
    return frac_last

print(lastplace(ballot_list, 'a'))
print(lastplace(ballot_list, 'b'))
print(lastplace(ballot_list, 'c'))
```

### But we could have also used our previous function!

```python
def plurality_last_fraction(bl):
    last_choice_votes=np.array(bl[:,-1])
    candidate_count= np.zeros(3)
    candidate_count[0]=len(last_choice_votes[last_choice_votes == 'a'])
    candidate_count[1]=len(last_choice_votes[last_choice_votes == 'b'])
    candidate_count[2]=len(last_choice_votes[last_choice_votes == 'c'])
    print("Last choice votes for each candidate:\n",candidate_count)
    return candidate_count[0]/len(bl), candidate_count[1]/len(bl), candidate_count[2]/len(bl)
```

```python
print("Last choice fractions?",plurality_last_fraction(ballot_list))
```
