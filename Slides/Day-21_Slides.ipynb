### What is "inheritance" and why would we use it?

Inheritance increases the **reusability** of our code and creates space for a hierarchy within classes (objects) we build.

- common "base" class: that contains a set of attributes and methods that all versions of that class should have,

- a "derived" class: for some use cases, it would be useful to have a few additional methods or attributes, 

- inheritance: "base" class + some added functionality that only makes sense for the new class.

The `super().__init__()` code is what ensures that our derived classes are initialized in the same way that the base class is initialized.

### Let's briefly revisit how this was done when created new "Animals"

```python
class Animal:
    """
        This is an Animal class, used to store information about animals in the Zoo.
        Or for characters in the Madagascar movies.
    """
    def __init__(self, kind, name, to_zoo='', area=100, staff=1):
        """ Initialize the Animal object.
        Inputs are
        kind : the type of animal (string)
        name : the nickname for the animal (string)
        to_zoo : the date the animal came to the zoo (in 'yyyy-mm-dd' format)
        """
        self.kind = kind
        self.name = name
        self.brought_to_zoo = to_zoo
        self.required_area = area  # required area [m2]
        self.required_staff = staff # required number of staff members per animal; can be fraction
```

```python
class Penguin(Animal):
    def __init__(self, name):
        """ Initialize Penguin class. All we need now is the animal nick name. """
        kind = 'penguin'
        # this calls initialization of Animal class and setting the animal to given 'name'
        super().__init__(kind, name)
        # in lines below we define attribute values for penguins
        self.required_area = 300
        self.required_staff = 0.5
```

### What is "composition" and why would we use it?

Composition: design principle where classes are built using instances of other classes, creating a "has-a" relationship instead of the "is-a" relationship typical of inheritance. 
- Includes **instances** of other classes as fields


The key advantages of composition are that it allows you to:
* Build complex systems from simpler components
* **Reuse** objects in different contexts
* Keep related data and functionality bundled together
* **flexibility**: With *inheritance*, changes in a base class affect all subclasses


### Let's take a look at how we can use `Animal` objects inside the `Zoo` methods.

### The extra methods from part 3.3/3.4:

```python
class Zoo:
    """
        Class for Zoo object.
    """
    def __init__(self, name):
        """ Initialize the Zoo object. """
        # name of the Zoo
        self.name = name
        # we are using dictionaries for animals and personnel
        self.animals = {}
        self.zookeepers = {}

    def add_animal(self, animal):
        """ Adds an animal to the zoo. """
        # Here the "animal" in the arguments refers to an Animal object,
        # so that we can use Animal's method such as .get_name()
        name = animal.get_name()
        # we use animal's name as a key in the dictionary, 
        # with Animal object itself as the value
        self.animals[name] = animal
```

#### Brief aside on Python dictionaries

```python
# Create a dictionary of animal space requirements
animal_space = {"lion": 100, "zebra": 50, "penguin": 20}

# I can add additional entries as needed
animal_space['monkey'] = 10

# Access a single value using its key
print(animal_space["lion"])  # prints: 100

# Get all keys
print("\nKeys:")
for animal in animal_space.keys():
    print(animal)

# Get all values 
print("\nValues:")
for space in animal_space.values():
    print(space)
```
