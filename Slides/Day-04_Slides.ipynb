## The importance of knowing if your code is "inside" or "outside" your conditional statement


```python
# initializing the list
greater_than_6_list = []
for i in range(11):
    print("i equals",i)
    if i > 6:
        print("i is greater than 6")
        # appending the number that is greater than 6
    greater_than_6_list.append(i) # I need to make sure this is "inside" the loop to do what I want
print("List of numbers to sum:",greater_than_6_list)
print("Sum of numbers is equal to:",sum(greater_than_6_list))
```

## Demystifying function creation

What are the important components needed when creating a function?

#### Let's list some important parts of writing functions:
* `def` -- defines the function
* name (comes right after def)
* arguments -- inside `()`
* `return` -- use this to get something back from the function
* It should serve a purpose

```python
# Calculating the hypotenuse of a triangle
a = 3
b = 4
hyp = (a**2 + b**2)**(1/2)
print(hyp)
```

```python
# Can we create a function that does this?
def get_h(a,b):
    hyp = (a**2 + b**2)**(1/2)
    return hyp
```

```python
## FOR INSTRUCTOR
# This function is written in class in real time
def get_h(a,b):
    h = (a**2 + b**2)**(1/2)
    return h
```

```python
# Let's try running it here for 5 and 12
result = get_h(5,12)
print(result)
```

## What if we wanted to return more than one thing?

```python
def get_h(a,b):
    h = (a**2 + b**2)**(1/2)
    h_times_2 = h * 2.0
    return h, h_times_2
```

```python
a = 3
b = 4
result, result_times_2 = get_h(a,b)
print(result, result_times_2)
```

## Variable scope (or "all the ways Python will try to drive me crazy")

#### First, let's look at the recommended approach

```python
def get_h(a,b):
    h = (a**2 + b**2)**(1/2)
    return h
```

```python
# Define my parameters/variables
my_a = 3
my_b = 4

#Call my function
result = get_h(my_a,my_b)
print(result)

# Will the code below work? If I haven't defined "h" anywhere outside my function, it won't work!
# If "h" is only defined inside the function, the rest of my code doesn't know about it.

# print(h)
```

#### Thought exercise: will this code work?

```python
a = 3
b = 4
def get_h():
    h = (a**2 + b**2)**(1/2)
    return h
result = get_h()
print(result)
```

#### Disclaimer: <font color="red">don't get in the habit of writing code like this!!</font>

```python
a = 3
b = 4
def get_h():
    h = (a**2 + b**2)**(1/2)
    return h
result = get_h()
print(result)
```

#### My code is no longer easily reusable later on!

```python
# Define new parameters/variables
new_a = 5
new_b = 12

#Call my function
result = get_h(new_a, new_b)
print(result)
```

### Also, when you use default argument values in your function, you can further confuse yourself if you're not being careful!

You'll see more about setting default arguments in the next PCA.

```python
a = 3
b = 4
def get_h(a=5, b=12):
    h = (a**2 + b**2)**(1/2)
    return h
result = get_h()
print(result)
```
