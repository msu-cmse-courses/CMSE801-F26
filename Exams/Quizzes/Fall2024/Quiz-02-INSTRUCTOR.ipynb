```python
# Will need to use NumPy for the function
import numpy as np
# Will need to import matplotlib for the plotting
import matplotlib.pyplot as plt

# This function calculates temperature over a range of days
def calculate_temperature(t, A, B, C, D):
     return A*np.cos(B*(t - C)) + D

# Create days for a full year
days = np.arange(0,365,1) + 1

### PUT YOUR CODE BELOW THIS POINT ###
# REMEMBER: you should be using the function calculate_temperature to calculate
#           the temperature for each day based on the model parameters and the provided "days" array

```

```python
### ANSWER
temps = calculate_temperature(days, 27.5, 0.017, 215, 57.5)
plt.plot(days,temps)
plt.xlabel('days')
plt.ylabel('temperature [F]')
plt.axvline(182,color="0.5", ls="--")
#plt.savefig("model_temps.png")

# 2 points for calling the function with the correct argument values
#    - take off a point if the function is called,
#       produces a result that can be used,
#       but is done so incorrectly
# 1 point for making the plot
# 1 point for labeling the x-axis
# 1 point for labeling the y-axis
# 1 point for putting the vertical line in the right place
```

## **3. Setting up the "derivatives" function for a system of ODEs (8 points)**

In this question, you're going to think about how one could build a model to study how the biomass (https://en.wikipedia.org/wiki/Biomass_(ecology)) in a forest changes as a function of time. We can model this system with the following equations:

$$ \frac{dx}{dt} = -x + 3y $$
$$ \frac{dy}{dt} = -3y + 5z $$
$$ \frac{dz}{dt} = -5z $$
