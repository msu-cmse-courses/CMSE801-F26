```python
### DO NOT MODIFY THIS CELL ###
days_of_week = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
hours_studied = [3, 2, 4, 3.5, 1, 0, 2]

index = 0
total_hours = 0

for day in days_of_week:
    print("On", day, "I studied", hours_studied[index], "hours.")
    total_hours += hours_studied[index]
    index += 1

print("In total, I studied",  total_hours, "hours this week.")
```
