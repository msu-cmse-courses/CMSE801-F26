## **2. Modeling a system using `solve_ivp` (13 points)**

&#9989;&nbsp;  **Question 2.1 (4 points)**: In this question, you're going return to the model for how the biomass in a forest changes as a function of time. Again, we'll be thinking about how biomass moves between living trees, dead trees, and humus, which is defined to be "the organic component of soil, formed by the decomposition of leaves and other plant material by soil microorganisms."

As before, we can model this system with the following equations (though note that the rate coefficients have been modified):

$$ \frac{dx}{dt} = -x + 1.4y $$
$$ \frac{dy}{dt} = -1.4y + 2.6z $$
$$ \frac{dz}{dt} = -2.6z $$

where $x$ is the biomass that has decayed in humus, $y$ is the biomass in dead trees, and $z$ is the biomass in living trees. Given that we're talking about the life and death of trees, we assume will use **decades** (10s of years) as our unit of time, $t$. This means that, for example, $t = 2$ would be _20 years_. 

Previously, you defined the derivatives function necessary to evolve this model, so now you're going to actually evolve the model using `solve_ivp` to evolve this model.

In the code cell below, **you are provided with a correct derivatives function for this model.**

Your task is to **solve this biomass system using `solve_ivp`**, assuming the following initial conditions:

* $x(0) = 0$
* $y(0) = 0$
* $z(0) = 809601$ in units of thousands of tons (based on 2012 data for forests in Michigan).

**Evolve the model for 5 decades (50 years) using a timestep of $\Delta t = 0.01$ decades.**

**Code you need to write to solve this problem**:
* A set of variables for storing your initial conditions, $x_0$, $y_0$, and $z_0$, set to the correct initial values.
* A variable that stores all the times for which you want to solve the system of ODEs, you can generate these times using `np.arange()`. The times should match the time range and timestep size specified above.
* Code that calls `solve_ivp` using all of the appropriate arguments and stores the result. 
* Code that unpacks the solution so that you can plot it.

**Note**: you will not need the "`args`" argument for `solve_ivp` for this case because any model parameters that might be needed for this system are provided directly as numbers in the set of  ODEs include above.

```python
#  DO NOT MODIFY THIS CELL -- Make sure to run this cell before you run the next cell though!

# These are the modules to need to import for this problem
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt

# This is the provide derivatives function for the ODE you're solving
def derivs(t, vals):
    x, y, z = vals
    dxdt = -x + 1.4*y
    dydt = -1.4*y + 2.6*z
    dzdt = -2.6*z
    return dxdt, dydt, dzdt
```

```python
### ANSWER

# Define your initial conditions
x0 = 0
y0 = 0
z0 = 809601
ICs = [x0, y0, z0]

# Define the time points you want to solve the ODE at
t = np.arange(0, 5, 0.01)

# Solve the ODE using solve_ivp
sol = solve_ivp(derivs, (t[0], t[-1]),ICs, t_eval=t)

# Unpack the solution
x = sol.y[0,:]
y = sol.y[1,:]
z = sol.y[2,:]
#print(x)

# Grading
# 1 point for defining the initial conditions correctly
# 1 point for defining the time points correctly
# 1 point for solving the ODE correctly using solve_ivp
# 1 point for unpacking the solution correctly
```
