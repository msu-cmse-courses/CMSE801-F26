Write a function called `lin_model` that computes a linear function of carat to predict price for just the `IF` diamonds. That is, the function to be learned and eventually passed to `curve_fit` is 
$$
\text{price} = m\cdot \text{carat} + b
$$


```python
# Put your code here
```

```python
### ANSWER ###
def lin_model(carat,m,b):
    return m*carat + b

# 0.5 point for correct name
# 1 point for correct order of the inputs. No partial credit
# 1 point for correct math. No partial credit
# 0.5 points for return statement
```

&#9989;&nbsp; **Question 2.2 (3 points):** 

- Use `curve_fit` to fit a linear model of price vs carat for just the `IF` diamonds.  
- What are the values of $m$ and $b$?

```python
# Put your code here
```

```python
### ANSWER ###

popt, pcov = curve_fit(lin_model, 
                       df_IF.carat, 
                       df_IF.price)


# 0.5 point for calling curve_fit
# 0.5 point for correct placement of lin_model
# 0.5 point for correct placement of y values (price)
# 0.5 point for correct placement of x values (carat)
# 0.5 point for assigning outputs (popt, pcov)

# print(popt)
print(f"m: {popt[0]}\nb: {popt[1]}")
# .5 point for printing m and b and having them labeled in the correct order
```
