## Part 2: Estimations of $\pi$ (5 points)

In this part, you will compute estimations of $\pi$. Knowing that the perimeter of a circle of radius 0.5 is equal to $\pi$, you could estimate its parameter and hence $\pi$ by the constructing regular polygons within it. This leads to the following recursive relation

\begin{equation*}
\large
p_{k+1} = 2^k\,\sqrt{ 2\Bigg(\,1-\sqrt{1-\Bigg(\frac{p_k}{2^k}\Bigg)^2}\,\Bigg)} 
\end{equation*}

for $k\geq2$ and where $p_k$ is the perimeter of a $2^k$-sides polygon. 

**Question 1 (3 points)**: Knowing that $p_2=2\sqrt{2}$, create a function `estimation_pi` that computes an approximation of $pi$. The input of this function is the parameter $k$ and the output is the estimation of $\pi$.

```python
# Put your code here
def estimation_pi(k):
    p = 2*np.sqrt(2)
    i = 2 
    while i<k:
        p = 2**i*np.sqrt( 2*(1-np.sqrt( 1 - (p/(2**i))**2)) )
        i += 1
    return p 
```

**Question 2 (2 points)**: Estimate $\pi$ using polygons with 4, 8, 16,$\dots$, 4096 sides. Verify your function by computing the error $\pi-p_k$. Print your results in the following format "`The approximation is ... with ... sides and its error is ...`"

```python
# Put your code here
for k in range(2,13):
    pi1 = estimation_pi(k)
    print("The approximation is ",pi1," with ", 2**k, "sides and its error is ", 
          np.pi-pi1)
```

## Part 4: Modeling a system using ODEINT (8 points)

In this question, you're going to think about how one could build a model to study algae blooms. In this section, we will model populations of plankton, which are organisms that cannot swim against the current. Phytoplankton (P) are plankton which get their nutrients from photosynthesis, while zooplankton (Z) are small protozoans or metazoans (e.g. crustaceans or other animals) that feed on other plankton. 
We will model this system with Lotka-Voltera model: 

$$ \frac{dP}{dt} = \alpha P - \beta P Z  $$
$$ \frac{dZ}{dt} = \delta PZ - \gamma Z$$

We assume will use **months** as our unit of time, $t$.

**Question 8 (2 points)**: Describe what the four parameters ($\alpha$, $\beta$, $\delta$ and $\gamma$) mean in the context of the biological application being modeled.

Put your answer here
* $\alpha$ is birth rate of P
* $\beta$ is rate that P gets eaten by Z 
* $\delta$ is rate of growth of Z  
* $\gamma$ is rate of death of Z

**Question 9 (3 points)**: Approximate the solution of this biomass system on a time interval of 12 months using `odeint` a timestep of $\Delta t = 0.01$. The initial conditions are given by

* $P(0) = 1$
* $Z(0) = 0.1$

and the parameters are
* $\alpha = 1$
* $\beta = 4/3$
* $\delta = 1$
* $\gamma = 1$

```python
def LV(current_state, time, alpha, beta, delta, gamma):
    
    p = current_state[0]
    z = current_state[1]
    
    dpdt = alpha*p - beta*p*z
    dzdt = delta*p*z - gamma*z
    
    return [dpdt, dzdt]

alpha = 1
beta = 4/3
delta = 1
gamma = 1
initial_conditions = [1,.1]
dt = 0.01
Nt = round(12/0.01+1)
time = np.linspace(0,12,Nt)

result = odeint(LV, initial_conditions, time, args=(alpha, beta, delta, gamma))

P = result[:,0]
Z = result[:,1]
```

**Question 10 (1 point)**: Make a single plot that displays the evolution of both variables as a function of time. Each line should be a different color and your plot should contain a $x$-axis label, a $y$-axis label, and a legend.

```python
plt.plot(time, P, label='Phytoplankton')
plt.plot(time, Z, label='Zooplankton')
plt.grid()
plt.legend()
plt.xlabel('time (months)')
plt.ylabel('population')
plt.title('Algae levels')
```

**Question 11 (2 points)**: **Without re-running your model**, make a second plot where you zoom in on just the first six months of the evolution of the system. Ensure that all of the same labels are included.

```python
### ANSWER ###
# Import needed modules

stop_index = int(len(time) *0.5)

plt.figure(1)
plt.plot(time[0:stop_index],P[0:stop_index],label="Phytoplankton")
plt.plot(time[0:stop_index],Z[0:stop_index],label="Zooplankton")
plt.grid()
plt.xlabel("Time (months)")
plt.ylabel("Population")
plt.legend()
```
