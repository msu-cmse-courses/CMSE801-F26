### ANSWER
The model above shows the most basic parts of the compartmental model for sharks and seals:
* Different compartments for sharks, seals, and fish
* Relationships that represent the fact that: sharks eat seals and fish, and seals eat fish
* The dotted green arrows are *not* indications of increasing/decreasing compartment sizes, they just point out which compartments influence the death rates of other compartments.

Many students may neglect to include fish, but you should **encourage them to add this compartment** so they can try to add a little complexity to their model.

This model is a great way to showcase the fact that compartments don't have to represent one population where individuals can move between compartments. * Birth rates and death rates for all of the animals

* An interaction where seals eat baby sharks (this actually happens!)

### ANSWER
The model above shows some parts of a rumor model that students might come up with:
* Different compartments for rumor spreaders (S), susceptible (U), exposed (E), and immune (I)
* A pathway from U to E, that depends on the size of S and U (i.e., rumors spread when spreaders expose people to the rumor)
* Pathways leading from E to S and E to I, indicating the when people are exposed to a rumor, they can either ignore it and become immune, and believe it and become a spreader
