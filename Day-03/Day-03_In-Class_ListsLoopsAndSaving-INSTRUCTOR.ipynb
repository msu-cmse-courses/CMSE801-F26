```python
### ANSWER ###


money_added = 2000.0 # we'll say they add 2000 dollars to their savings every year 
interest_rate = 0.065
years_of_investment = 65-22

total_savings = []
current_savings = 0
old_savings = 0

for jj in range(years_of_investment):
    current_savings = old_savings + old_savings*interest_rate + money_added
    total_savings.append(current_savings)
    old_savings = current_savings
```

**Take the code you wrote in Part 2 and expand on it so that you can calculate total retirement savings for a range of different input parameters.** To start, try 3 different annual contribution values and 4 different interests rates, but try to write your code so that you could easily do greater or fewer numbers of values for both parameters.

```python
### ANSWER ###

money_added_list = [2500,5000.0,10000.0]
interest_rate_list = [0.025, 0.05, 0.075, 0.1]
years_of_investment = 65-22

for pp in range(len(money_added_list)):
    money_added = money_added_list[pp]
    
    for ii in range(len(interest_rate_list)):
        interest_rate = interest_rate_list[ii]
        total_savings = []
        current_savings = 0
        old_savings = 0
        
        for jj in range(years_of_investment):
            current_savings = old_savings + old_savings*interest_rate + money_added
            total_savings.append(current_savings)
            old_savings = current_savings

        print("For annual money added of ",money_added,", and interest rate of ",interest_rate, ", final savings were ", current_savings/1.0e6, " (in millions)")
```
