### Let's practice some of these concepts with a simple class
**Remember**: A class is made of three main things: name, attributes, and methods

```python
class Student():
    '''
    This class is designed to include information about individual students.
    Currently this class has the following attributes:
    
    name : this is the student's name
    gpa : this is the student's current gpa
    year : this is what year of college the student is in
    '''
    def __init__(self, name='', gpa=0.0, year=1):
        self.name = name
        self.gpa = gpa
        self.year = year
```

```python
theo = Student('Theodore', gpa=3.2)
```

```python
print(theo.name)
print(theo.gpa)
print(theo.year)
```

#### Now we'll expand that class a bit with some methods.

```python
class Student():
    '''
    This class is designed to include information about individual students.
    
    There are now also methods for interacting with the Student class.
    '''
    def __init__(self, name='', gpa=0.0, year=1):
        self.name = name
        self.gpa = gpa
        self.year = year
        self.courses = []
    
    def get_name(self):
        print("My name is %s." %self.name)
        
    def enroll(self,courses):
        self.courses = courses
        
    def display_enrolled(self):
        if len(self.courses) == 0:
            print("I am not enrolled in any courses at the moment")
        else:
            print("I am enrolled in %s and %s." %(self.courses[0], self.courses[1]))
        
    def years_until_graduation(self):
        return 4 - self.year
```

####  And then we can test those new methods

```python
theo = Student('Theodore', gpa=3.2)
```

```python
theo.get_name()
```

```python
theo.enroll(['CMSE 201', 'MTH 133'])
theo.display_enrolled()
```

```python
theo.years_until_graduation()
```
