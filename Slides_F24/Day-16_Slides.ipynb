## Let's look at a some ways we can use seaborn

```python
# Some some imports out of the way
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Let's make some data to plot
# Create a list from 0 to 2pi with increments of 0.1
x_start = 0
x_stop = 2*np.pi
dx = 0.1
x = np.arange(x_start, x_stop, dx)

# Calculate sine and cosine using NumPy
sin_x = np.sin(x)
cos_x = np.cos(x)
```

```python
# Turn on the seaborn styling
sns.set_theme()

plt.plot(x,sin_x)
plt.plot(x,cos_x)
```

```python
# We can also modify the style; details here: https://seaborn.pydata.org/tutorial/aesthetics.html
sns.set_style("ticks") # we can test "ticks", "dark", "whitegrid" as well

plt.plot(x,sin_x)
plt.plot(x,cos_x)
```

```python
# We can also modify the style; details here: https://seaborn.pydata.org/tutorial/aesthetics.html
sns.set_style("ticks")

plt.plot(x,sin_x)
plt.plot(x,cos_x)

# we can remove the "spines" as well
sns.despine()
```

```python
# We can start getting really specific as well
sns.set_style("darkgrid", {"axes.facecolor": ".9"})

plt.plot(x,sin_x)
plt.plot(x,cos_x)
```

```python
# We can see all the things we can tweak with this command:
sns.axes_style()
```

```python
# And finally, we can set a "context" for our plot
sns.set_context("talk")

sns.set_style("white")

plt.plot(x,sin_x)
plt.plot(x,cos_x)

sns.despine()
```
