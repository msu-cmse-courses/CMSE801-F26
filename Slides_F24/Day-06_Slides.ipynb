### Generating a set of finer-grained x values and plotting sine and cosine

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a list from 0 to 2pi with increments of 0.1
x_list = []
x = 0
dx = 0.1
while x < 2*np.pi:
    x_list.append(x)
    x += dx

# Calculate sine and cosine element-by-elements
sin_xlist = []
cos_xlist = []
for x in x_list:
    sin_xlist.append(np.sin(x))
    cos_xlist.append(np.cos(x))
    
# Calculating it for the whole list as once (numpy let's us do this!)
sin_xlist = np.sin(x_list)
cos_xlist = np.cos(x_list)
```

```python
# I could have used a NumPy function to create the x values (you'll see this in the next PCA)
dx = 0.1
x_list = np.arange(0., 2.0 * np.pi, dx)
print(x_list)
```

```python
# Plotting
plt.plot(x_list,sin_xlist,"g",label='sin(x)') # I general try to avoid being lazy about just
                                              # using "g" instead of using color="g"
plt.plot(x_list,cos_xlist,color="orange", label="cos(x)")
plt.legend()
plt.xlabel("x")
plt.ylabel("f(x)");
```

#### I can also get really carried away with customizing my plots

```python
# Define some parameters for the plot to customize it, note that this changes the defaults for all future plots
params = {"ytick.color" : "w",
          "xtick.color" : "w",
          "axes.labelcolor" : "w",
          "axes.edgecolor" : "w",
          "axes.facecolor" : "#18453B",
          "legend.facecolor" : "w"}
plt.rcParams.update(params)

# # Plotting
plt.figure(facecolor = "#18453B", figsize = (3.8, 2.7) )
plt.plot(x_list,sin_xlist,"g",label='sin(x)') # Note that a green line isn't maybe the best choice with our spartan-themed plot
plt.plot(x_list,cos_xlist,color="orange", label="cos(x)")
plt.legend()
plt.xlabel("x")
plt.ylabel("f(x)");
```
