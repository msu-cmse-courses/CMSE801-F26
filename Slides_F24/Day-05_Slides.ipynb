## Let's look at some of the things we can do with functions by taking the simple "Happy Birthday" example


```python
# Define a function that prints "Happy Birthday to <person>" where the person is a name arguement I pass in
def happy_birthday(name):
    print("Happy birthday to", name)
happy_birthday("Ian")
```

```python
# What if I want there to be a default name?
def happy_birthday(name='you!'):
    print("Happy birthday to", name)
happy_birthday()
```

```python
# What if I want to return the statement instead of just print it?
def happy_birthday(name='you!'):
    return("Happy birthday to"+" "+name)
wishes = happy_birthday("Sparty") 
print(wishes)
#happy_birthday("Ian")
```

**Do the following**:

1. Write function called `x_squared`.
2. The function should take in a list of $x$-values of any length, but should have a default of [1,2,3].
3. The function should **return** a new list of values that are $x^2$ values of the original input values.
4. Print the resulting new list.

```python
def x_squared(x_vals=[1,2,3]):
    x2 = []
    for x in x_vals:
        x2.append(x**2)
    return x2
```

```python
# Solution for instructor reference
def x_squared(vals=[1,2,3]):
    x2 = [] # Initialize the list to return
    for x in vals: # Looping by ???
        x2.append(x**2)
    return x2

xsquared = x_squared([2,4,6,8,10])
print(xsquared)
```

```python
# Let's test it out
xs = [2,4,6,8,10]

squared = x_squared(xs)
print(squared)

# Let's also confirm that my default ([1,2,3]) is working:
default_squared = x_squared()
print (default_squared)
```

### What if we wanted to use a different function _inside_ this function?

For example, I could use my `happy_birthday` function from before (though it might not be the obvious choice...)

```python
def x_squared(vals=[1,2,3]):
    x2 = [] # Initialize the list to return
    for x in vals: # Looping by ???
        x2.append(x**2)
        if x**2 > 50:
            print(happy_birthday("someone!"))
    return x2

squared = x_squared([2,4,6,8,10])
print(squared)
    
```

```python
# I could also pass a default value for when I call the happy_birthday function
def x_squared(vals=[1,2,3], hb='someone else!'):
    x2 = [] # Initialize the list to return
    for x in vals: # Looping by ???
        x2.append(x**2)
        if x**2 > 50:
            print(happy_birthday(hb))
    return x2

squared = x_squared([2,4,6,8,10])
print(squared)
```
