### Let's try loading up the data from the PCA

For reference, we may want to take a look at the [NumPy documentation](https://numpy.org/doc/stable/reference/generated/numpy.loadtxt.html).

```python
# Import numpy before I try to use it!
import numpy as np

# Load the data using NumPy's loadtxt function
pop_dates, pop_numbers = np.loadtxt("michigan_pop.csv", usecols = (0,1), unpack=True, skiprows = 1, delimiter=',')
```

What do the following arugments do?
* `usecols`: tells NumPy which columns of the dataset to load
* `skiprows`: tells NumPy if it should skip any of the rows and, if so, how many to skip
* `delimiter`: tells NumPy what character(s) are used to separate columns in the file
* `unpack`: splits the columns into separate variables

### Let's take a peek at our data

```python
# Let's look at just the first 5 values in each of my new variable
print("First five dates:", pop_dates[0:5])
print("First five population values:", pop_numbers[0:5])
```

```python
# Or I could have used a loop!
for i in range(5):
    print(pop_dates[i], pop_numbers[i])
```

```python
# What if I don't use "unpack"
pop_dates, pop_numbers = np.loadtxt("michigan_pop.csv", usecols = (0,1), skiprows = 1, delimiter=',')
```

```python
pop_values = np.loadtxt("michigan_pop.csv", usecols = (0,1), skiprows = 1, delimiter=',')
```

```python
# Do I still have all the same values?
print(pop_values[0:5,:]) # note that for 2D arrays, the first index is for rows, second index is columns
```

### Let's take a look at this data since we've gone through the trouble of loading it up

```python
# Need to import before I can make plots!
import matplotlib.pyplot as plt

plt.plot(pop_dates, pop_numbers/1e6)
plt.xlabel('Year')
plt.ylabel('Population [millions]');
```

#### What if I needed to add some more values? (important to check the [documentation](https://numpy.org/doc/stable/reference/generated/numpy.append.html))

```python
# appending to an array (different than how strings work!)
# be mindful of the choice to overwrite the original data
pop_dates = np.append(pop_dates,[2021, 2022, 2023, 2024])
print(pop_dates)

# And I could do the same thing with the "pop_numbers" array!
```
