```python
# import NumPy
import numpy as np

def assign_groups(models, nGroups):
    # Create a dictionary for storing group assignments
    assignments = {}
    # Loop through the number of groups and pick a random model
    # Print out the group assignments once a choice is made
    for i in range(nGroups):
        assigned = np.random.choice(list(models.values()))
        assignments[i] = assigned
        print("Group %i is assigned the %s model. \n" %(i+1,assigned))
    return assignments
    
# Define the model options
models = {1:"SIR",2:"LV",3:"DK"}

# How many groups are there?
nGroups = 10
    
# Check to see if there are at least three of each model
seek_solution = True
while seek_solution:
    
    # Call the assign groups function
    assignments = assign_groups(models, nGroups)
    
    # Do the group checking
    sir = 0
    lv = 0
    dk = 0
    for key in assignments.keys():
        if assignments[key] == "SIR":
            sir += 1
        if assignments[key] == "LV":
            lv += 1
        if assignments[key] == "DK":
            dk += 1
    if sir >= 3 and lv >= 3 and dk >= 3:
        seek_solution = False
        print("Found a solution! \n")
    else:
        print(">>> Not enough of each model, try again! <<< \n") 
```
