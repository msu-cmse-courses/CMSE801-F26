```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
%matplotlib inline

# Note that the SciPy library is needed for odeint
from scipy.integrate import odeint
```

```python
def derivs(y, t, beta, delta, p, r):
    
    # unpack list of incoming variables
    target = y[0]
    infected = y[1]
    virion = y[2]
    
    dTdt = -beta*target*virion
    didt = (beta*target*virion) - (delta*infected)
    dVdt = (p*infected) - (r*virion)
    
    return [dTdt, didt, dVdt]
```

```python
x = np.linspace(0,2*np.pi,50)
y = np.sin(x)
z = np.cos(x)
```

```python
fig = plt.figure(figsize=(15,5))
ax1 = fig.add_subplot(1,2,1)
ax1.plot(x, y)
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')
ax2 = fig.add_subplot(1,2,2, projection = '3d')
ax2.plot(x, y, z)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('z')
plt.tight_layout()
```

```python
fig = plt.figure(figsize=(15,5))
ax = fig.add_subplot(projection = '3d')
ax.plot(x, y, z)
ax.view_init(10,20)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.tight_layout()
```
