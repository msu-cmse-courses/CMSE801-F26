```python
class Athlete():
    def __init__(self, name, sport, next_training, next_competition='NA'):
        ''' Initialize the object Building. '''
        self.name = name
        self.sport = sport
        self.next_training = next_training
        self.next_competition = next_competition
        return
    def set_next_competition(self, next_competition):
        self.next_competition = next_competition
        return
    def get_sport(self):
        return self.sport
    def get_info(self):
        s = 'Name: '+self.name+' Sport: '+self.sport+' N.T.: '+self.next_training+' N.C.: '+self.next_competition
        return s
```

```python
class Boxer(Athlete):
    def __init__(self,name,next_training,numberMatches=0):
        super().__init__(name,'boxe',next_training)
        self.numberMatches = numberMatches
        return
    def get_numberMatches(self):
        return self.numberMatches
```
