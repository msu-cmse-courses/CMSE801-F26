## Order of Magnitude Modeling

### How much time do I spend driving my car per day?

### Assumptions

* only go to work 
* only take highways

### Inputs
* distance between work and home: $l$ in km
* speed: $v$ in km/h 

### Output
* time: t in hours

### Ranges 
* $l \in [100, 110]$
* $v \in [100, 120]$
* $t \in [t_{min}, t_{max}]$

### Equations
$$t = 2\frac{l}{v}$$

$$t_{max} = 2 \frac{l_{max}}{v_{min}}$$
$$t_{min} = 2 \frac{l_{min}}{v_{max}}$$

```python
lmin = 100
lmax = 110
vmin = 100
vmax = 120
tmin = 2*lmin/vmax
tmax = 2*lmax/vmin
print("t is between ", tmin, " hours and ", tmax, "hours." )
```
