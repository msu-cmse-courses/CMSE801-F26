```python
def create_forest(N, ftreestart):
    
    # Initialize a 2D array full of zeros using np.zeros
    forest = np.zeros((N,N))

    # Loop through the 2D array to initialize the trees    
    for i in range(N):
        for j in range(N):
            if np.random.random()< ftreestart:
                forest[i,j] = 1
    
    forest[:,0] = 2
    
    return forest
```

```python
forest_array = create_forest(5, .3)
print(forest_array)
```
