```python
# we import math module
import math 

# define x, sin(x), cos(x)
xList = list()
sinList = list()
cosList = list()
x = 0
dx = 0.1
while x < 2*math.pi:
    xList.append(x)
    sinList.append(math.sin(x))
    cosList.append(math.cos(x))
    x += dx 
```

```python
print('x = ', xList)
print('sin(x) = ', sinList)
print('cos(x) = ', cosList)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.plot(xList,sinList,"g",label='sin(x)',marker="*",linestyle="-")
plt.plot(xList,cosList,color="orange", label="cos(x)",linestyle="-.")
plt.legend()
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Graph") 
plt.show()
```

```python
plt.figure(figsize=[15,2])
plt.plot(xList,sinList,"g",marker="*",linestyle="-")
plt.xlabel("x")
plt.ylabel("sin(x)")

plt.figure(figsize=[15,2])
plt.plot(xList,cosList,color="orange",linestyle="-.")
plt.xlabel("x")
plt.ylabel("cos(x)")
```

```python
figure, axes = plt.subplots(2,1,figsize=[12,3])
axes[0].plot(xList,sinList,color="g")
axes[0].set_xlabel('x')
axes[0].set_ylabel('sin(x)')
axes[1].plot(xList,cosList,color="orange")
axes[1].set_xlabel('x')
axes[1].set_ylabel('cos(x)')
```
