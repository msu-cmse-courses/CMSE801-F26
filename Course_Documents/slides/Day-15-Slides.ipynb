```python
import pandas as pd 
set_example = pd.DataFrame([ ['A',1,2], ['B', 3, 4], ['C', 5, 6]])
print(set_example)
```

```python
set_example.index = set_example[0]
print(set_example)
set_example = set_example.iloc[:,1:]
print(set_example)
set_example.index.name = None
print(set_example)
set_example.columns = ['c1', 'c2']
print(set_example)
```

```python
print(set_example.iloc[0,0])
print("---")
print(set_example.iloc[1:3,0])
print("---")
print(set_example.iloc[1,0:2])
```

```python
print(set_example)
print(set_example['c1'].loc['A'])
print(set_example.loc['A','c1'])
print(type(set_example.loc['A','c1']))
print("---")
print(set_example.loc['B':'C','c1'])
print(type(set_example.loc['B':'C','c1']))
print("---")
print(set_example.loc['B','c1':'c2'])
```

```python
mask = set_example < 4
print(mask)
print(set_example[mask])
set_example[mask].describe()
```
