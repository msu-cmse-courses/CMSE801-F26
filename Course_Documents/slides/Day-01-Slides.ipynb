## Three Threads of CMSE 801

### 1. Coding in Python 

By the end of CMSE 801, you will be a proficient programmer/coder in Python. You should be able to use Python in all future classes.

### 2. Modeling

You will learn how to build models that allow you to quantitatively solve complex problems in science, engineering and beyond. Combined with the coding aspect of the course, these models will be turned into computational models.

### 3. Data Analysis

You will learn how to handle real-world data: find it, analyze it, and draw quantitative conclusions from it. You will use your coding skills for this analysis and build models that reveal deeper insights into the data.
