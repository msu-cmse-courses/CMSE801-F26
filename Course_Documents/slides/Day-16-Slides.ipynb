```python
import seaborn as sns
v = [0.0, 0.25, 0.5, 0.75, 1]
w = [0.0, 0.25, 0.5, 0.75, 1]
fig = sns.jointplot(x=v, y=w)
fig.ax_joint.set_xlabel('x')
fig.ax_joint.set_ylabel('y')
fig.fig.suptitle('y vs x')
```
