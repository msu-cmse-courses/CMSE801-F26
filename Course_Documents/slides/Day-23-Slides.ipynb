## Steps to convert your Jupyter notebook into a PDF file

* In the top menu of your Jupyter notebook, click on Kernel > Restart & Run All 


* In the top menu of you Jupyter notebook, click on File > Print Preview. A new window in your default web browser should open.


* Print the preview displayed in your web browser and save it as a PDF (here I use Google Chrome).
