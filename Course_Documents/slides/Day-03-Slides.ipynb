```python
# for loop by index
grade = [9, 10, 7, 8, 9]
info = list()
for i in range(len(Name)):
    student_info = list()
    student_info.append(Name[i])
    student_info.append(grade[i])
    info.append(student_info)
print(info)
print(info[0])
print(info[0][1])
```

```python
# for loop by value
print(grade)
average = 0
for i in grade:
    print(i)
    average += i # average = average + i
average /= len(grade) # average = average/len(grade)
print(average)
```

```python
for i in info:
    print(i)
```

```python
for i in info:
    print(i[0], 'obtains', i[1])
```

```python
# while loop
student = 'Law'
sdum = ' '
i = -1 
while sdum != student:
    i += 1
    sdum = info[i][0]
print(i)
print(student, 'obtains', info[i][1])
```

```python
x = [2,4,6,8,10,12,14,16,18]
x2 = list()
for i in x:
    x2.append(i**2)
print(x2)
```

```python
# nested loop 
print(info)
for i in range(4):
    print('---')
    print('i = ',i)
    print('---')
    for j in range(2):
        print('   j = ',j)
        print('      info[i][j] =', info[i][j])
```
