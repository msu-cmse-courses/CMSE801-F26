```python
# Import modules
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np 
```

```python
x_list = ['allo', 34, 3.1]
x_array = np.array(['allo', 34, 3.1])

print(x_list)
print(x_array)
```

```python
x = np.arange(0,1.25,0.25)
y = np.linspace(0,1.25,3)
print("x = ", x)
print(len(x))
print("y = ", y)
print(len(y))
print("second component of x :",x[1])
print("sum of x :", x.sum)
print("mean of x :", x.mean())
```

```python
y = [ [1, 2, 3], [4, 5, 6]]
print(type(y))
y = np.array([ [1, 2, 3], [4, 5, 6]])
print(type(y))
print(y)
print(y[:,1])
print(y[0,:])
print(y[1,2])
```

```python
x = np.zeros((2,3))
print(x)
x = np.ones((5,4))
print(x)
```

```python
x = np.array([[0, 1, 2],[4, 5, 8]])
y = np.ones_like(x)
print(y)
print(x+y)
print(np.cos(x))

print(x)
print(x+1)
```
