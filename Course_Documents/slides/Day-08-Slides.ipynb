# Initial value problem
\begin{equation*}
\begin{aligned}
    y'(t) =&\,\, -5\,y(t) \\
    y(0) =&\,\, 1.1
\end{aligned}
\end{equation*}

We want an estimation of $y(3)$.

\begin{equation*}
    y'(t_n) \approx \frac{y(t_{n+1})-y(t_n)}{t_{n+1}-t_n} = \frac{y(t_{n}+\Delta t)-y(t_n)}{\Delta t}
\end{equation*}

\begin{equation*}
\begin{aligned}
    \frac{y_{n+1}-y_n}{\Delta t} =&\,\, -5\,y_n \\
    y_{n+1} =&\,\, y_n - 5\,\Delta t\,y_n \\
    y_{\text{new}} =&\,\, y_{\text{old}} - 5\,\Delta t\, y_{\text{old}}
\end{aligned}
\end{equation*}
   
If we known the initial condition $y_0 = 1.1$ (or $y_{\text{old}} =1.1$), we can find an estimation of $y(3)$!

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Derivative function
def derivs(curr_vals, time):
    dydt = -5*curr_vals
    return dydt
```

```python
# initial condition
y0 = 1.1
# Define the time array
t0 = 0.0
tmax = 3.0
dt = 0.01
time = np.arange(t0, tmax + dt, dt)

# Solve the odes with odeint
sol = odeint(derivs, y0, time)
```

```python
plt.figure()
plt.plot(time, sol,color="green")
plt.xlabel('t')
plt.ylabel('y')
plt.grid()
```

```python
# pre-class examples

# Derivative function
def derivs(curr_vals, time):
    
    # Declare parameters
    g = -9.81 # m/s^2
    
    # Unpack the current values of the variables we wish 
    # to "update" from the curr_vals list
    h, v = curr_vals
    
    # Right-hand side of odes, which are used to computer the derivative
    dhdt = v
    dvdt = g
    
    return dhdt, dvdt
```

```python
# Declare Variables for initial conditions
h0 = 3000 # meters
v0 = 0 # m/s
tmax = 10 # seconds
dt = 1 # seconds

# Define the time array
time = np.arange(0, tmax + dt, dt)

# Store the initial values in a list
# WARNING: it must be consistent with the 
#          function derivs
init = [h0, v0]  

# Solve the odes with odeint
sol = odeint(derivs, init, time)
```

```python
plt.figure
# Plot the height using the "0" element from the solution
plt.plot(time, sol[:,0],color="green")
# Plot the velocity using the "1" element from the solution
plt.plot(time, sol[:,1],color="purple")
plt.xlabel('Time [s]')
plt.grid()
```
