```python
import numpy as np

x = np.random.rand(5,5)
print(x)
```

```python
# first column
print(x[:,0])
# second row from component 2 to the end
print(x[1,1:])
# second row from component 2 to 3
print(x[1,1:3])
print(x[1:3,1:3])
```

```python
# copy 
y = x.copy()
x[0] = 2.0
print(y)
print(x)
```
