```python
def get_stat_exam(exam_info):
    nb_absence = 0
    absence_list = list()
    nb_success = 0
    nb_attend = 0
    mean = 0 
    for i in exam_info:
        if i[1] > 6:
            nb_success += 1
            nb_attend += 1
            mean += i[1]
        elif i[1] == -1:
            nb_absence += 1
            absence_list.append(i[0])
        else:
            nb_attend +=1 
            mean += i[1]
    mean /= nb_attend
    return mean, nb_success, nb_absence, absence_list
```

```python
mean, nb_success, nb_absence, absence_list = get_stat_exam(info)
print(mean, ' ', nb_success, ' ', nb_absence, ' ', absence_list)
```
