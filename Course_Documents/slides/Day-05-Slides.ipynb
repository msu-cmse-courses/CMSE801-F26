```python
def exam_data(grades):
    nbStudents = len(grades)
    if nbStudents == 0:
        print("There is no data")
    else:
        mean = sum(grades)/nbStudents
    print('mean = ', mean)
    return mean, nbStudents

# mean, nbStudents  = exam_data([1, 2, 3, 4, 5, 6, 7])
exam_data([1, 2, 3, 4, 5, 6, 7])
print(mean, nbStudents)
exam_data([])
```

```python
def print_grades(grades):
    for i in grades:
        print(i)

print_grades([1, 2, 3, 4, 5, 6, 7])
```

```python
def x_squared(vals=[1,2,3]):
    x2 = []
    for x in vals:
        x2.append(x**2)
    return x2

xsquared = x_squared([2,4,6,8,10])
print(xsquared)
```
